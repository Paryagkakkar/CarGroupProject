```python
# ──────────────────────────────────────────────────────────────────────────────
#          AMAZON SUPPLY CHAIN INTELLIGENCE – Delivery Delay Risk Prediction
# ──────────────────────────────────────────────────────────────────────────────
# Group Project – AI/ML Course
# Objective: Multi-class classification → On-Time | At Risk | Delayed
# Business Value: Proactive alerts → reroute / reassign / notify customers
# ──────────────────────────────────────────────────────────────────────────────

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from math import radians, sin, cos, sqrt, atan2
from datetime import datetime, time
import zipfile

from sklearn.model_selection import train_test_split, StratifiedKFold, GridSearchCV
from sklearn.preprocessing import LabelEncoder, StandardScaler, OrdinalEncoder, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.metrics import (classification_report, accuracy_score, f1_score,
                             confusion_matrix, make_scorer)
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.dummy import DummyClassifier
from xgboost import XGBClassifier

from imblearn.over_sampling import SMOTE
from imblearn.pipeline import Pipeline as ImbPipeline
from imblearn.under_sampling import RandomUnderSampler

import warnings
warnings.filterwarnings('ignore')

sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

# ─── 1. Load & Initial Cleaning ──────────────────────────────────────────────

# Adjust path if needed (Colab / local / Kaggle)
with zipfile.ZipFile('amazon_delivery.zip', 'r') as zip_ref:
    zip_ref.extractall('.')
df = pd.read_csv('amazon_delivery.csv', encoding='latin1', on_bad_lines='skip')   # or full path to unzipped file

print("Shape:", df.shape)
print(df.info())

# Clean categorical columns
cat_cols = ['Weather', 'Traffic', 'Vehicle', 'Area', 'Category']
for col in cat_cols:
    if col in df.columns:
        df[col] = df[col].str.strip().replace('', np.nan)

# Convert times
df['Order_Time']  = pd.to_datetime(df['Order_Time'],  format='%H:%M:%S', errors='coerce').fillna(pd.to_datetime('00:00:00')).dt.time
df['Pickup_Time'] = pd.to_datetime(df['Pickup_Time'], format='%H:%M:%S', errors='coerce').fillna(pd.to_datetime('00:00:00')).dt.time

# ─── 2. Feature Engineering ──────────────────────────────────────────────────

def haversine(lat1, lon1, lat2, lon2):
    R = 6371  # Earth radius in km
    dlat = radians(lat2 - lat1)
    dlon = radians(lon2 - lon1)
    a = sin(dlat/2)**2 + cos(radians(lat1)) * cos(radians(lat2)) * sin(dlon/2)**2
    c = 2 * atan2(sqrt(a), sqrt(1-a))
    return R * c

df['Distance_km'] = df.apply(lambda r: haversine(r['Store_Latitude'], r['Store_Longitude'],
                                                 r['Drop_Latitude'],  r['Drop_Longitude']), axis=1)

# Processing time (pickup - order) in minutes – handle overnight edge cases
def minutes_between(t1, t2):
    h1, m1, s1 = t1.hour, t1.minute, t1.second
    h2, m2, s2 = t2.hour, t2.minute, t2.second
    delta = (h2*3600 + m2*60 + s2) - (h1*3600 + m1*60 + s1)
    return delta / 60 if delta >= 0 else (delta + 86400) / 60  # +24h if negative

df['Processing_min'] = df.apply(lambda r: minutes_between(r['Order_Time'], r['Pickup_Time']), axis=1)

# Time-based features
df['Order_Hour']   = df['Order_Time'].apply(lambda t: t.hour)
df['Order_DayOfWeek'] = pd.to_datetime(df['Order_Date']).dt.dayofweek
df['Is_Weekend']   = df['Order_DayOfWeek'].isin([5,6]).astype(int)

# Agent reliability proxy (lower age + higher rating = better?)
df['Agent_Reliability'] = df['Agent_Rating'] / (df['Agent_Age'] + 1)  # avoid div by zero

# ─── Risk Scoring (Innovation) ───────────────────────────────────────────────

weather_risk = {'Sunny':0, 'Cloudy':1, 'Windy':2, 'Fog':3, 'Stormy':4, 'Sandstorms':5}
traffic_risk = {'Low':0, 'Medium':1, 'High':2, 'Jam':3}
vehicle_risk = {'bicycle':3, 'scooter':2, 'motorcycle':1}  # bicycle slowest/riskiest?

df['Weather_Risk']  = df['Weather'].map(weather_risk).fillna(2)   # median impute
df['Traffic_Risk']  = df['Traffic'].map(traffic_risk).fillna(1)
df['Vehicle_Risk']  = df['Vehicle'].map(vehicle_risk).fillna(2)

df['Composite_Risk'] = (
    0.30 * df['Weather_Risk'] +
    0.30 * df['Traffic_Risk'] +
    0.15 * df['Vehicle_Risk'] +
    0.15 * (df['Distance_km'] / df['Distance_km'].quantile(0.95)).clip(0,3) +  # cap extreme
    0.10 * (1 - df['Agent_Reliability'].clip(0,1))                             # poor agent → higher risk
)

# Interaction feature example
df['Risk_Distance'] = df['Composite_Risk'] * np.log1p(df['Distance_km'])

# ─── Target Creation (more realistic bins) ───────────────────────────────────

# Use approximate business logic or quantiles
print(df['Delivery_Time'].describe())

# Example: < 45 min = On-Time, 45–150 = At Risk, >150 = Delayed
# Adjust based on your data percentiles or stakeholder input
bins = [0, 45, 150, np.inf]
labels = ['On-Time', 'At Risk', 'Delayed']
df['Risk_Level'] = pd.cut(df['Delivery_Time'], bins=bins, labels=labels, include_lowest=True)

print("\nTarget Distribution:")
print(df['Risk_Level'].value_counts(normalize=True))

sns.countplot(data=df, x='Risk_Level', order=labels)
plt.title("Delay Risk Class Distribution")
plt.show()

# ─── Prepare Data ─────────────────────────────────────────────────────────────

drop_cols = ['Order_ID', 'Order_Date', 'Order_Time', 'Pickup_Time', 'Delivery_Time',
             'Store_Latitude','Store_Longitude','Drop_Latitude','Drop_Longitude']

X = df.drop(columns=drop_cols + ['Risk_Level'])
y_raw = df['Risk_Level']

le = LabelEncoder()
y = le.fit_transform(y_raw)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# ─── Preprocessing Pipeline ──────────────────────────────────────────────────

num_features = ['Agent_Age', 'Agent_Rating', 'Distance_km', 'Processing_min',
                'Order_Hour', 'Order_DayOfWeek', 'Is_Weekend', 'Agent_Reliability',
                'Composite_Risk', 'Risk_Distance']

cat_features = ['Weather', 'Traffic', 'Vehicle', 'Area', 'Category']

preprocessor = ColumnTransformer([
    ('num', Pipeline([
        ('impute', SimpleImputer(strategy='median')),
        ('scale',  StandardScaler())
    ]), num_features),

    ('cat', Pipeline([
        ('impute', SimpleImputer(strategy='most_frequent')),
        ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
    ]), cat_features)
], remainder='drop')

# ─── Models & Evaluation Function ────────────────────────────────────────────

def evaluate_model(name, pipeline, X_tr, y_tr, X_te, y_te):
    pipeline.fit(X_tr, y_tr)
    y_pred = pipeline.predict(X_te)

    acc  = accuracy_score(y_te, y_pred)
    f1_w = f1_score(y_te, y_pred, average='weighted')
    f1_m = f1_score(y_te, y_pred, average='macro')

    print(f"\n=== {name} ===")
    print(f"Accuracy : {acc:.4f}   |   F1 Weighted : {f1_w:.4f}   |   F1 Macro : {f1_m:.4f}")
    print(classification_report(y_te, y_pred, target_names=le.classes_))

    # Confusion Matrix
    cm = confusion_matrix(y_te, y_pred)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=le.classes_, yticklabels=le.classes_)
    plt.title(f"{name} – Confusion Matrix")
    plt.xlabel("Predicted"); plt.ylabel("True")
    plt.show()

    return pipeline, acc, f1_w, f1_m

# ─── Model Comparison ────────────────────────────────────────────────────────

models = {
    "LogisticRegression": LogisticRegression(max_iter=2000, class_weight='balanced'),
    "RandomForest": RandomForestClassifier(random_state=42, class_weight='balanced_subsample'),
    "XGBoost": XGBClassifier(random_state=42, eval_metric='mlogloss', scale_pos_weight=1),  # we'll tune later
    "GradientBoosting": GradientBoostingClassifier(random_state=42)
}

results = []

for name, clf in models.items():
    pipe = ImbPipeline([
        ('prep', preprocessor),
        ('smote', SMOTE(random_state=42, sampling_strategy='auto')),
        ('clf',   clf)
    ])

    fitted_pipe, acc, f1w, f1m = evaluate_model(name, pipe, X_train, y_train, X_test, y_test)
    results.append({'Model': name, 'Acc': acc, 'F1_w': f1w, 'F1_m': f1m, 'Pipe': fitted_pipe})

# ─── Feature Importance (from best tree-based model) ─────────────────────────

best_model_name = max(results, key=lambda x: x['F1_w'])['Model']
best_pipe = next(r['Pipe'] for r in results if r['Model'] == best_model_name)

if best_model_name in ["RandomForest", "XGBoost", "GradientBoosting"]:
    clf_step = best_pipe.named_steps['clf']
    importances = clf_step.feature_importances_

    feature_names = (best_pipe.named_steps['prep']
                     .get_feature_names_out())

    fi_df = pd.DataFrame({'Feature': feature_names, 'Importance': importances})
    fi_df = fi_df.sort_values('Importance', ascending=False).head(20)

    plt.figure(figsize=(10,8))
    sns.barplot(data=fi_df, x='Importance', y='Feature')
    plt.title(f"{name} – Confusion Matrix")
    plt.xlabel("Predicted"); plt.ylabel("True")
    plt.show()

# ─── Quick Hyperparameter Tuning Example (XGBoost) ───────────────────────────

xgb_pipe = ImbPipeline([
    ('prep', preprocessor),
    ('smote', SMOTE(random_state=42)),
    ('xgb', XGBClassifier(random_state=42, eval_metric='mlogloss'))
])

param_grid = {
    'xgb__n_estimators': [100, 200],
    'xgb__max_depth': [4, 6, 8],
    'xgb__learning_rate': [0.05, 0.1]
}

grid = GridSearchCV(xgb_pipe, param_grid, cv=3, scoring='f1_weighted', n_jobs=-1)
grid.fit(X_train, y_train)

print("Best XGBoost Params:", grid.best_params_)
evaluate_model("Tuned XGBoost", grid.best_estimator_, X_train, y_train, X_test, y_test)

# ─── Save Best Model (Deployment Ready) ──────────────────────────────────────

import joblib
joblib.dump(grid.best_estimator_, 'amazon_delay_risk_model.pkl')
print("Model saved → amazon_delay_risk_model.pkl")

# ─── Summary Table ───────────────────────────────────────────────────────────

res_df = pd.DataFrame(results)[['Model','Acc','F1_w','F1_m']]
res_df = res_df.sort_values('F1_w', ascending=False)
print("\nFinal Model Comparison:")
print(res_df.round(4))

sns.barplot(data=res_df.melt(id_vars='Model'), x='Model', y='value', hue='variable')
plt.title("Model Comparison – Accuracy vs F1 Scores")
plt.xticks(rotation=45)
plt.show()

# ─── Business Recommendations ────────────────────────────────────────────────
"""
Top Insights:
• Composite_Risk, Distance_km, Traffic_Risk usually rank high → real-time traffic/weather integration would help
• At Risk class is hardest → consider cost-sensitive learning or threshold tuning
• Possible Action: For 'At Risk' predictions → auto-notify customer / assign faster vehicle

Future Work:
• Live API integration (weather + traffic)
• SHAP explanations per prediction
• Streamlit / FastAPI deployment
"""
```
